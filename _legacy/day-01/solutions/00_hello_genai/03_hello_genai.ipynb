### Step 1: Set up the environment 

```python
import sys
sys.path.append('../../../')

from dotenv import load_dotenv, find_dotenv
from litellm import completion, get_valid_models
from lib.validation import validate_environment_for_model
import requests
import os

# Load the environment
load_dotenv(find_dotenv())
```

### Step 2: Show available models
In order to get a list of available models, make sure that you set the base-url and api-key (if needed) in *.env* for your provider correctly (see *README.md*).

```python
# Get the available models for providers. Use "gemini", "ollama" or "litellm_proxy" - this depends on your available infrastructure.
PROVIDER = "ollama"

# Call the endpoint for getting 
valid_models = get_valid_models(
    check_provider_endpoint=True,
    custom_llm_provider=PROVIDER
)

# Print the available models
for model in valid_models:
    print(model)
```

```python
# BTW: This is the similar to calling the respective endpoint directly
ollama_base_url = os.getenv("OLLAMA_API_BASE")
requests.get(ollama_base_url + "/v1/models").json()
```

### Step 3: Configure the GenAI model 

```python
# Use one of the models above. Unfortunately, there is no endpoint/method for checking the task a models is used for (generation, embedding, reranker), so you have search the provider documentation (or ask the instructor).
MODEL = "deepseek-r1:1.5b"
MODEL_ID = f"{PROVIDER}/{MODEL}"
```

```python
# Check if all necessary environment variables are set.
validate_environment_for_model(model=MODEL_ID)
```

### Step 4: Set up the prompt 

```python
SYSTEM_PROMPT = "You are a friendly assistant with a preference for London."
USER_PROMPT = "What is the most beautiful city in the world?"
```

### Step 5: Setup the generation configuration

```python
# Set some generation parameters. Later, we will learn more about them.
TEMPERATURE = 1.2
MAX_TOKENS = 800
```

```python
# Print out summary of input values / parameters 
print(f'SYSTEM PROMPT used:\n {SYSTEM_PROMPT}')
print(f'USER PROMPT used:\n {USER_PROMPT}')
print(f'MODEL used:\n {MODEL} (temperature = {TEMPERATURE}, max_output_tokens = {MAX_TOKENS})')
```

### Exercise 01: Call the model
Generate response by calling the *completion* method from `litellm` (see imports). Use the *model*, *messages* and *generation configuration parameters* as arguments.

```python
# TODO: Set up the messages aka. provide the system prompt for role "system" and the user prompt for the role "user"
messages = [
        {"role":"system", "content": SYSTEM_PROMPT},
        {"role": "user", "content": USER_PROMPT}
]

# TODO: Generate response by calling the "completion" method from litellm.
# Use the model, messages and generation configuration parameters as arguments.
response = completion(
    model=MODEL_ID,
    messages=messages,
    temperature=TEMPERATURE,
    max_tokens=MAX_TOKENS
)
```

```python
# Print the complete response of genai model 
print('RESPONSE of genAI model:')
print(response)
```

### Exercise 02: Print the message of the genAI model
Print only the message of the genAI model. What is the `role` of the model?

```python
# TODO: Print the role of the message
print('ROLE of the genAI model:')
print(response.choices[0].message.role)
print()

# TODO: Print the message
print('MESSAGE of genAI model:')
print(response.choices[0].message.content)
```

### Exercise 03: Set default model in *.env*
You should set a **default model for generation** in order to not having to define this within each notebook again. For this follow the steps:  
1. Choose a model, which successfully generated a response (i.e. `gpt-oss:20b`)
2. Update the variable `DEFAULT_GENERATION_MODEL` of your *.env*-file located at the root of this repository
3. Restart the kernel of this notebook by clicking *Restart* above this notebook
4. Execute the following cells

```python
# We need to redefine the imports since we restartet the kernel
from dotenv import load_dotenv, find_dotenv
from litellm import completion
import os

# Load the environment
load_dotenv(find_dotenv())

# Print the default generation model. This should be the model you selected
DEFAULT_GENERATION_PROVIDER = os.getenv("DEFAULT_GENERATION_PROVIDER")
default_generation_model= os.getenv("DEFAULT_GENERATION_MODEL")

print(f"Your default model provider: {DEFAULT_GENERATION_PROVIDER}")
print(f"Your default model for generating text: {default_generation_model}")
```

```python
# This format is needed by litellm
default_generation_model_id = f"{DEFAULT_GENERATION_PROVIDER}/{default_generation_model}"

# Prepare a message
messages = [
        {"role": "user", "content": "Say 'Hello World!'"}
]

# Call the completion for this model
response = completion(
    model=default_generation_model_id,
    messages=messages,
)

# Print the response. If this results in 'Hello World!", you set up the default generation model correctly!
print(f"Response: {response.choices[0].message.content}")
```
