#### Joining strings (concatenation)

There are several ways to put strings together:

* `+` joins two strings. Only strings can be joined this way – convert numbers with `str(...)`.
* `+=` appends to an existing string – useful in loops.
* **f-strings** (an `f` before the quotes) insert values in curly braces: `f"Ticket {ticket_id}"`. That's the way you'll use most. Behind a colon you can format numbers: `{costs:.4f}` shows 4 decimal places.
* `"separator".join(list)` joins a list of strings with a separator, e.g. one per line with `"\n"`.

`\n` inside a string is a **line break**.

````python
ticket_id = "WORK-412"
status = "open"
costs = 0.0031746

# With +
print("Ticket " + ticket_id + " is " + status)
print("Number of tickets: " + str(9))

# With an f-string
print(f"Ticket {ticket_id} is {status}")
print(f"Costs: {costs:.4f} USD")

# With += in a loop
index = ""
for name in ["write-adr", "write-commit-message"]:
    index += f"- {name}\n"
print(index)

# With join
lines = ["first line", "second line", "third line"]
print("\n".join(lines))
````

### 4. Lists

A **list** holds several values in an order. You write it in square brackets.

* Access by position (the **index**), starting at `0`. `[-1]` is the last element.
* `append` adds an element at the end, `len` counts the elements, `in` checks whether an element is in the list.

````python
ticket_ids = ["WORK-412", "WORK-421", "WORK-423"]

print(ticket_ids[0])        # the first element
print(ticket_ids[-1])       # the last element
print(ticket_ids[:2])       # the first two elements
print(len(ticket_ids))      # 3

ticket_ids.append("WORK-430")
print(ticket_ids)
print("WORK-421" in ticket_ids)
````

### 5. Dictionaries

A **dictionary** stores values under a **key** – like a ticket with its fields. You write it in curly braces: `{"key": value}`.

* `ticket["status"]` reads a value. If the key doesn't exist, you get a `KeyError`.
* `ticket.get("assignee")` reads a value and returns `None` if the key doesn't exist.
* `ticket["status"] = "done"` changes or adds a value.

````python
ticket = {
    "id": "WORK-412",
    "title": "500 Internal Server Error when creating a booking without timezone",
    "status": "open",
}

print(ticket["id"])
print(ticket.get("assignee"))       # None - the key doesn't exist
ticket["status"] = "in progress"
print(ticket)
````

#### Lists of dictionaries: the messages of a conversation

The most important data structure of this workshop: the conversation with a model is a **list of dictionaries**. Each message has a `role` (`system`, `user`, `assistant` or `tool`) and a `content`.

````python
messages = [
    {"role": "system", "content": "You are a helpful coding assistant."},
    {"role": "user", "content": "What does HTTP status 422 mean?"},
]

# The answer of the model is appended - the conversation grows
messages.append({"role": "assistant", "content": "The request is valid, but violates a business rule."})

print(len(messages))
print(messages[-1]["content"])      # the content of the last message
````

### 7. Loops

A `for` loop runs the indented code once for every element.

* `for item in list:` – over the elements of a list
* `for i in range(3):` – three times, with `i` = 0, 1, 2
* `for key, value in dictionary.items():` – over the entries of a dictionary

````python
tickets = [
    {"id": "WORK-412", "status": "open"},
    {"id": "WORK-415", "status": "done"},
    {"id": "WORK-421", "status": "open"},
]

# Find all open tickets
open_tickets = []
for ticket in tickets:
    if ticket["status"] == "open":
        open_tickets.append(ticket["id"])
print(open_tickets)

for step in range(1, 4):
    print(f"Step {step}")

for key, value in tickets[0].items():
    print(f"{key}: {value}")
````

**List comprehensions** are a short form of such a loop. You don't have to write them yourself, but you'll see them in the workshop code. This line creates the same list as the loop above:

````python
open_tickets = [ticket["id"] for ticket in tickets if ticket["status"] == "open"]
print(open_tickets)
````

### 8. Functions

A **function** is a named piece of code that you can call again and again. It gets values (**parameters**) and gives back a result with `return`.

* Parameters can have a **default value** (`top_k: int = 3`) – then you don't have to pass them.
* When calling, you can name the parameters: `search("rules", top_k=5)`.
* **Type hints** (`query: str`, `-> str`) say which types a function expects and returns.
* The **docstring** (the string in three quotes directly below `def`) describes the function.

Type hints and docstrings are especially important in this workshop: frameworks like FastMCP (MCP lesson) and Google ADK (agents lesson) turn them into the **tool description** for the model. For a model, a docstring is a prompt.

````python
def get_ticket(ticket_id: str, tickets: list) -> str:
    """Returns the title of a ticket.

    Args:
        ticket_id: The id of the ticket, e.g. "WORK-412".
        tickets: The list of all tickets.
    """
    for ticket in tickets:
        if ticket["id"] == ticket_id:
            return ticket["title"]
    return f"Error: the ticket {ticket_id} does not exist."


tickets = [{"id": "WORK-412", "title": "500 error without timezone"}]
print(get_ticket("WORK-412", tickets))
print(get_ticket("WORK-999", tickets=tickets))
````

````python
def count_words(text: str, minimum_length: int = 1) -> int:
    """Counts the words of a text that have at least minimum_length characters."""
    count = 0
    for word in text.split():
        if len(word) >= minimum_length:
            count += 1
    return count


print(count_words("Rooms can be booked for at most four hours"))
print(count_words("Rooms can be booked for at most four hours", minimum_length=5))
````

### 9. JSON

**JSON** is the text format in which programs exchange data – also models: the arguments of a tool call arrive as a JSON **string**. The module `json` converts between strings and Python values:

* `json.loads(text)` – JSON string → dictionary / list
* `json.dumps(value)` – dictionary / list → JSON string (`indent=2` makes it readable)

````python
import json

# This is how the arguments of a tool call arrive: as a string
arguments = '{"query": "maximum booking duration", "top_k": 3}'

data = json.loads(arguments)
print(data["query"])

print(json.dumps(data, indent=2))
````

### 11. `await`

Some functions wait for something, e.g. for the answer of an MCP server. In Python they are called **asynchronous**, and you call them with `await` in front. In a notebook you can simply write `result = await some_function(...)` – read it as "wait for the result". You don't have to write such functions yourself.

````python
import asyncio


async def ask_server(question: str) -> str:
    await asyncio.sleep(1)  # pretend to wait for a server
    return f"Answer to: {question}"


answer = await ask_server("Which tools do you offer?")
print(answer)
````

### 12. Pydantic

[Pydantic](https://docs.pydantic.dev/) describes the **shape of data** with a class: which fields, which types. We use it when the model has to answer in a fixed structure (*structured output*, prompting lesson) – e.g. a code review with a list of findings.

* A model is a class that inherits from `BaseModel`. Each field has a name and a type.
* `Field(description=...)` adds a description – the model sees it.
* `model_json_schema()` shows the JSON schema that is sent to the model.
* `model_validate_json(text)` reads the JSON answer of the model and **checks** it: missing fields or wrong types raise an error.
* You access the values as attributes: `review.findings[0].rule`.

````python
from pydantic import BaseModel, Field


class Finding(BaseModel):
    rule: str = Field(description="The id of the violated rule, e.g. SEC-04")
    explanation: str


class CodeReview(BaseModel):
    approved: bool
    findings: list[Finding]


print(json.dumps(CodeReview.model_json_schema(), indent=2))
````

````python
# The answer of a model - a JSON string
answer = '{"approved": false, "findings": [{"rule": "SEC-04", "explanation": "The meeting title is logged."}]}'

review = CodeReview.model_validate_json(answer)
print(review.approved)
print(review.findings[0].rule)
print(review.findings[0].explanation)
````

````python
# You can also create an object yourself
finding = Finding(rule="API-07", explanation="Timestamps without offset must be rejected.")
print(finding)
print(finding.model_dump())  # back to a dictionary
````

### 13. Reading error messages

When something goes wrong, Python shows a **traceback**. Start reading at the **last line**: it names the type of the error and the reason. The lines above show where it happened.

```
Traceback (most recent call last):
  Cell In[3], line 2
    print(ticket["assignee"])
KeyError: 'assignee'
```

| Error | Typical reason |
|---|---|
| `NameError: name 'x' is not defined` | a cell above was not run, or a typo in the name |
| `KeyError: 'assignee'` | the dictionary has no such key – check the spelling or use `.get()` |
| `IndexError: list index out of range` | the list has fewer elements than you think |
| `TypeError: can only concatenate str (not "int") to str` | `+` with a string and a number – use `str(...)` or an f-string |
| `IndentationError` / `SyntaxError` | wrong indentation, a missing `:` or bracket |
| `AttributeError: 'NoneType' object has no attribute ...` | a value is `None` – e.g. the model returned no text |
